# Cost of Living Targets

```python
from dash import Dash, html

import components as comp
from lve_plan import LvePlan
from lve_projector import LveProjector
from schemas import LvePlanID
from tax_profiler import TaxProfiler


# Parameters
plan_key = LvePlanID.MAIN
proj_len = 3
gst_rate = 0.05
base_ror = 0.09
base_cpi = 0.04
self_emp = False


# Data Objects
lve_plan = LvePlan(plan_key)
tax_profiler = TaxProfiler(self_emp)
lve_projector = LveProjector(lve_plan, proj_len, gst_rate)
lve_by_year = lve_projector.lve_by_year


# View Generation
app = Dash()
app.layout = [
    html.Div(
        [
            comp.lve_totals_card(lve_by_year),
            comp.lvi_totals_card(lve_by_year, tax_profiler),
            comp.lps_targets_card(lve_by_year, base_ror, base_cpi, tax_profiler),
            comp.lve_allocation_card(lve_by_year),
            comp.dsc_summary_card(lve_plan),
        ],
        style=comp.BASE_APP_STYLES,
    )
]
app.run(jupyter_mode="inline", jupyter_width="100%", jupyter_height=1200)
```
